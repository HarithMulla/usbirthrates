```python
# factorial
m=int(input("enter a num"))
mul=1
for i in range(1,m+1):
    mul=mul*i
print("fact is",mul)
```

```python
# harmonic num
n=int(input("enter a number:"))
if n<=0:
    print("invalid num")
else:
    harm=0
    for i in range(1,n):
        harm+=1/i
    print(harm)
```

```python
year=int(input("enter year"))
if year%4==0  and year%100!=0:
    print(year,"It is a leap year")
elif year%400==0:
    print(year,"it is leap year")
else:
    print(year," is not a leap year")
```
