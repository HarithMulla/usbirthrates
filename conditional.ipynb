```python
# divisible by 3 and5

num = int(input("enter a number"))

if num % 3==0 and num%5==0:
    print("num is divisible by both")
elif num % 3==0:
    print("num is only divisible by 3")
elif num % 5==0:
    print("num is only divisible by 5")
else:
    print("not divisible by both")
```
