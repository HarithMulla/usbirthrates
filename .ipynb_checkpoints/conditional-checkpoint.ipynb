```python
# even or odd 
num= int(input("enter a number"))

if num % 2 ==0:
    print("num is even")

else:
    print("num is odd")
```

```python
# pos and neg num 
num = int (input("enter a number"))
if num>0:
    print('num is positive')
elif num<0:
    print('numis negative')


else:
    print('zero')
```
