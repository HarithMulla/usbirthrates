# while loops

# In Python, a while loop is used to execute a block of statements repeatedly until a given condition is satisfied. When the condition becomes false, the line immediately after the loop in the program is executed.

```python
n=int(input("enter a num :"))
count=0
while(count<n):
    count=count+1          
    print("Harith")
```

# Infinite While Loop

# If we want a block of code to execute infinite number of times then we can use the while loop in Python to do so.
# "True" condition is that the loop will run infinitely
# for termination we use "break" keyword or other logic

```python
for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()
```

# Break and Continue Statement

#The continue statement is used to skip the current iteration of a loop and move to the next iteration. It is useful when we want to bypass certain conditions without terminating the loop.

```python
for letters in "Assalamualaikum":
    if letters=='a' or letters=='l':
        continue
    print('present letter',letters)
    
```

#break statement is used to exit the loop prematurely when a specified condition is met.

```python
for letters in "Assalamualaikum":
    if letters=='a':
        break
    print('present letter',letters)
```
