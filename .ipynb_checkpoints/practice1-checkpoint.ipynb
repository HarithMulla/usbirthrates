```python
# sum of n numbers
n=int(input("enter a number:"))
addition=0
for i in range(0,n+1):
    addition=addition+i
print( "sum of n num is:",addition)


```

```python
# factorial
m=int(input("enter a num"))
mul=1
for i in range(1,m+1):
    mul=mul*i
print("fact is",mul)
```

```python
# harmonic num
n=int(input("enter a number:"))
if n<=0:
    print("invalid num")
else:
    harm=0
    for i in range(1,n):
        harm+=1/i
    print(harm)
```
